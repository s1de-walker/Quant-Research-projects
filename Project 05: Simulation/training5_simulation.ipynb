## Simulation

```python
import numpy as np
import matplotlib.pyplot as plt

n_trades = 100

pnl = np.random.choice(
    [1, -1],
    size=n_trades,
    p=[0.55, 0.45]
)

cum_pnl = np.cumsum(pnl)

plt.figure(figsize=(10,5))
plt.plot(cum_pnl)
plt.axhline(0, linestyle='--')
plt.title("Biased Coin Trading Strategy")
plt.xlabel("Trade Number")
plt.ylabel("Cumulative PnL")
plt.show()

print("Final PnL:", pnl.sum())
```

```python
import numpy as np

n_sims = 10000
n_trades = 100

results = []

for _ in range(n_sims):

    pnl = np.random.choice(
        [1,-1],
        size=n_trades,
        p=[0.55,0.45]
    )

    results.append(pnl.sum())

print(results)    


plt.hist(results, bins=50)
plt.title("Distribution of Strategy Outcomes")
plt.xlabel("PnL")
plt.ylabel("Frequency")
plt.show()

print("Mean:", np.mean(results))
print("Std:", np.std(results))
```
