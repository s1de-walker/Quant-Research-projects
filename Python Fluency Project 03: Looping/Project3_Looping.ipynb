# Part 1: Loop Fundamentals | for, range(), and len(df)

```python
import pandas as pd

df = pd.DataFrame({
    "Stock": ["AAPL", "MSFT", "NVDA", "AMZN",
              "GOOG", "META", "TSLA", "AVGO"],
    "Price": [220, 510, 180, 225, 190, 700, 330, 300],
    "Volume": [1000, 1500, 2400, 1800, 900, 2100, 3000, 1600]
})

df
```

### Method A — **`len(df)`** and **`.iloc[i]`**

```python
for i in range(len(df)):
    print(i)
```

```python
for i in range(len(df)):
    print(df["Stock"].iloc[i])
```

```python
for i in range(len(df)):
    stock = df.iloc[i]["Stock"]
    price = df.iloc[i]["Price"]

    print(stock, price)
```

`len(df)`

Returns 8, because there are eight rows.


`range(8)`

Produces 0, 1, 2, 3, 4, 5, 6, 7.


`iloc[i]`

Fetches the row at that numerical position.

```python
for i in range(len(df)):
    stock = df["Stock"].iloc[i]
    price = df["Price"].iloc[i]

    print(stock, price)
```

### Method B: `df.iterrows()`

```python
for index, row in df.iterrows():
    print(row["Stock"], row["Price"])
```

# Part 2: Accessing different rows inside a loop

```python
import pandas as pd

df = pd.DataFrame({
    "Date": pd.date_range("2026-01-01", periods=8),
    "Price": [100, 104, 102, 108, 105, 112, 110, 118],
    "Volume": [1000, 1200, 900, 1500, 1100, 1800, 1400, 2000]
})

df
```

```python
# 1. Current row
for i in range(len(df)):
    current_price = df.iloc[i]["Price"]
    print(current_price)
```

```python
# 2. Previous row
# Start from i = 1, because the first row has no previous row.

for i in range(1, len(df)):
    current_price = df.iloc[i]["Price"]
    
    print(current_price)
```

```python
# 3. Two rows earlier
# Start from i = 2, because you need two earlier rows to exist.

for i in range(2, len(df)):
    tminus3_price = df.iloc[i]["Price"]

    print(tminus3_price)
```

**Start at n if you want to go back n days**

```python
# 3. Two rows earlier
# Start from i = 2, because you need two earlier rows to exist. 2-2 will take you to 0.

for i in range(2, len(df)):
    current_price = df.iloc[i]["Price"]
    tm2_price = df.iloc[i-2]["Price"]

    print(current_price, tm2_price)
```

```python
# 1. Calculate the percentage price change between the current row and the previous row.

# Formula: (current_price / previous_price) - 1

for i in range (1, len(df)):
    t = df.iloc[i]["Price"]
    tm1 = df.iloc[i - 1]["Price"]
    ret = t / tm1 - 1

    print(ret)
```

```python
# 2. Print the current price and the volume from two rows earlier.

for i in range (2, len(df)):
    price = df.iloc[i]["Price"]
    volume = df.iloc[i-2]["Volume"]

    print(price, volume)
```

```python
# 3. A strategy buys when today's price is greater than yesterday's price AND yesterday's volume was above 1,000. 
# Write the loop condition to check this. You do not need to implement the actual buying.

for i in range(1, len(df)):
    d = df.iloc[i]["Date"]
    t = df.iloc[i]["Price"]
    tm1 = df.iloc[i-1]["Price"]
    vol_m_1 = df.iloc[i-1]["Volume"]

    if t > tm1 and vol_m_1 > 1000:
        print(d, t, tm1, vol_m_1, "buy")
```

```python
# 4. Calculate the three-row price change:

for i in range (3, len(df)):
    d = df.iloc[i]["Date"]
    current_price = df.iloc[i]["Price"]
    tm3_price = df.iloc[i-3]["Price"]

    ret_3d = current_price / tm3_price - 1

    print(d, ret_3d)
```

# Part 4: Nested loops

```python
import pandas as pd

df = pd.DataFrame({
    "Date": pd.date_range("2026-01-01", periods=4),
    "AAPL": [100, 103, 101, 105],
    "MSFT": [200, 198, 202, 205],
    "NVDA": [150, 155, 153, 160]
})

df["Date"] = pd.to_datetime(df["Date"])
df = df.set_index("Date")

df
```

```python
### Part 1: Get the columns in a list
stocks = df.columns
results = []

stocks
```

```python
### Part 2: Run the outer loop and access in inner loop

for stock in stocks:
    for i in range(1, len(df)):

        d = df.index[i]
        current_price = df.iloc[i][stock]
        previous_price = df.iloc[i - 1][stock]

        signal = 1  if current_price > previous_price else 0

        results.append([d, stock, signal])


results
```

```python
### Part 3: Build the results df
results_df = pd.DataFrame(
    results,
    columns=["Date", "Stock", "Signal"]
)

results_df
```

```python
### Part 4: Pivot back to wide format
signal_df = results_df.pivot(index = "Date", columns= "Stock", values= "Signal")

signal_df
```

```python
### Full Structure

# Part 1: Get the columns in a list
stocks = df.columns
results = []

# Part 2: Run the outer loop and access in inner loop

for stock in stocks:
    for i in range(1, len(df)):

        d = df.index[i]
        current_price = df.iloc[i][stock]
        previous_price = df.iloc[i - 1][stock]

        signal = 1  if current_price > previous_price else 0

        results.append([d, stock, signal])

# Part 3: Build the results df
results_df = pd.DataFrame(
    results,
    columns=["Date", "Stock", "Signal"]
)

# Part 4: Pivot back to wide format
signal_df = results_df.pivot(index = "Date", columns= "Stock", values= "Signal")

signal_df
```

# Part 5: Handling data in loops

### Handling missing data

```python
df = pd.DataFrame({
    "Date": pd.date_range("2026-01-01", periods=5),
    "AAPL": [100, 105, None, 110, 115],
    "MSFT": [200, 198, 202, None, 205]
}).set_index("Date")

df
```

```python
# Skip invalid observations using continue

results = []

for stock in df.columns:
    for i in range(1,len(df)):
        d = df.index[i]
        current_price = df.iloc[i][stock]
        previous_price = df.iloc[i-1][stock]

        if pd.isna(current_price) or pd.isna(previous_price):
            continue

        ret = current_price / previous_price - 1

        results.append([d, stock, ret])

results_df = pd.DataFrame(results, columns=["Date", "Stock", "Returns"])

results_df
```

```python
results_df_wide = results_df.pivot(index="Date", columns="Stock", values="Returns")
results_df_wide
```

### Different amounts of historical data

```python
df = pd.DataFrame({
    "Date": pd.date_range("2026-01-01", periods=5),
    "AAPL": [100, 102, 104, 103, 106],
    "NEW": [None, None, 50, 52, 55]
}).set_index("Date")

df
```

```python
results = []

for stock in df.columns:
    valid_prices = df[stock].dropna()

    if len(valid_prices) < 3:
        continue

    for i in range(1, len(df)):
        d = df.index[i]
        current_price = df.iloc[i][stock]
        previous_price = df.iloc[i-1][stock]

        if pd.isna(current_price) or pd.isna(previous_price):
            continue

        ret = current_price / previous_price - 1
        results.append([d, stock, ret])

results_df = pd.DataFrame(results, columns=["Date", "Stock", "Returns"])

results_df
```

```python
### Practise
import yfinance as yf
import pandas as pd

tickers = ["IXN", "IXC", "URTH"]

# Fetch a buffer of calendar days, then keep the last 50 trading days
raw = yf.download(tickers, period="1y", auto_adjust=True, progress=False)
prices = raw["Close"].dropna(how="all").tail(250)
prices.tail()
```

Skip stocks with fewer than 100 valid prices.

Calculate the 5-day return for each eligible stock.

Store the results in a DataFrame with Date, Stock, and Return_5D.

```python
results = []

for stock in prices.columns:

    valid_prices = prices[stock].dropna()

    if len(valid_prices) < 100:
        continue

    for i in range(5, len(prices)):
        d = prices.index[i]
        current_px = prices.iloc[i][stock]
        px_5d = prices.iloc[i-5][stock]

        if pd.isna(current_px) or pd.isna(px_5d):
            continue

        ret_5d = current_px / px_5d - 1

        results.append([d, stock, ret_5d])


results_df = pd.DataFrame(results, columns=["Date", "Stock", " Returns_5d"])

results_df.tail()
results_df.isna().sum()
```
