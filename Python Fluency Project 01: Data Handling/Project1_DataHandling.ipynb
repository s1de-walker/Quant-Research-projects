## **Part 1: loc vs iloc**

```python
df = pd.DataFrame({
    "Ticker": ["AAPL", "MSFT", "GOOG", "AMZN"],
    "Price": [200, 420, 175, 190],
    "Return": [0.02, -0.01, 0.03, 0.01]
})
```

```python
df
```

Rule  
loc = labels  
iloc = integer positions

```python
df.loc[1, "Price"]
```

row label 1, column label "Price"

```python
df.iloc[1, 1]
```

second row, second column

#### Slicing

iloc excludes the last number index

```python
df.iloc[1:3]
```

returns rows 1,2  
excudes the last number index

**but loc behaves differently**  

it includes the last number index

```python
df.loc[1:3]
```

**Important point**  
>iloc slicing excludes the endpoint.  
>loc slicing includes the endpoint.

### **Understanding the comma**  
df.iloc[rows, columns]

before comma = rows  
after comma = columns

```python
df.iloc[0:2, 1:3]
```

rows 0, 1  
columns 1,2 

### **Understanding :**
: means everything

```python
df.iloc[:, 1]
```

all rows, column 1

```python
df.iloc[1, :]
```

row 1, all columns 

**Important point**  

**df.iloc[ROWS, COLUMNS]**

### **Boolean filtering**

```python
df.loc[df["Return"] > 0]
```

can select columns inside a **list**

```python
df.loc[df["Return"] > 0, ["Ticker", "Return"]]
```

# **Part 4: Combining datasets**

```python
prices = pd.DataFrame({
    "Ticker": ["AAPL", "MSFT", "NVDA", "AAPL"],
    "Date": ["2026-01-01", "2026-01-01",
             "2026-01-01", "2026-01-02"],
    "Price": [250, 420, 140, 255]
})

prices
```

```python
weights = pd.DataFrame({
    "Ticker": ["AAPL", "MSFT", "GOOGL"],
    "Weight": [0.4, 0.35, 0.25]
})

weights
```

### **merge() — combine using matching columns**

```python
result = prices.merge(weights, on="Ticker", how="left")

result
```

- **on**="Ticker" — the **common column** used to match rows.

- **how**="left" — keep every row from prices.

- GOOGL doesn't appear in the output because it isn't present in prices.

### **join() — combine using the index**

```python
prices_by_ticker = prices.set_index("Ticker")
weights_by_ticker = weights.set_index("Ticker")

result = prices_by_ticker.join(weights_by_ticker,how="left")

result
```

The key difference:
>- merge() usually matches using columns.
>- join() usually matches using the index of each DataFrame.

### **concat() — stack or place datasets together**

```python
jan_prices = pd.DataFrame({
    "Ticker": ["AAPL", "MSFT"],
    "Price": [250, 420]
})

feb_prices = pd.DataFrame({
    "Ticker": ["AAPL", "MSFT"],
    "Price": [255, 430]
})
```

```python
jan_prices
```

```python
feb_prices
```

```python
all_prices = pd.concat(
    [jan_prices, feb_prices],
    axis=0,
    ignore_index=True
)
```

```python
all_prices
```

```python
import pandas as pd

prices = pd.DataFrame({
    "Date": ["2026-01-01", "2026-01-01", "2026-01-02", "2026-01-02"],
    "Ticker": ["AAPL", "MSFT", "AAPL", "MSFT"],
    "Price": [250, 420, 255, 425]
})

weights = pd.DataFrame({
    "Ticker": ["AAPL", "MSFT"],
    "Weight": [0.6, 0.4]
})

factors = pd.DataFrame({
    "Date": ["2026-01-01", "2026-01-01", "2026-01-02"],
    "Ticker": ["AAPL", "MSFT", "AAPL"],
    "Momentum": [0.8, 0.3, 0.9]
})
```

```python
# Q1. Merge prices with weights using Ticker. Keep every price row.

result = prices.merge(weights, on ="Ticker", how = "left")

result
```

```python
# Q2. Merge the result with factors using both Date and Ticker. Keep every price row.

result2 = result.merge(factors, on = ["Date" , "Ticker"], how = "left")

result2
```

#### indicator — find unmatched records

```python
result3 = result.merge(
    factors,
    on=["Date", "Ticker"],
    how="left",
    indicator=True
)

result3
```

```python
result3[result3["_merge"] != "both"]
```

# **Part 6: Reshaping Data**

```python
import pandas as pd

df = pd.DataFrame({
    "Date": ["Day1", "Day1", "Day2", "Day2"],
    "Ticker": ["AAPL", "MSFT", "AAPL", "MSFT"],
    "Price": [250, 420, 255, 425]
})

df
```

```python
wide = df.pivot(index="Date", columns="Ticker", values="Price")

wide
```

```python
# how to remove duplicate values for the same index?
wide.index.duplicated(keep='first')                         # what u want to remove
wide = wide[~wide.index.duplicated(keep='first')]


# OR

wide = wide.groupby(level=0).mean()    # or .last(), .first(), .sum(), .median()
```

# **Part 7: Data manipulation tools**

```python
df = pd.DataFrame({
    "Ticker": ["AAPL", "MSFT", "NVDA", "AAPL", "MSFT", "NVDA"],
    "Sector": ["Tech", "Tech", "Tech", "Tech", "Tech", "Tech"],
    "Date": ["Day1", "Day1", "Day1", "Day2", "Day2", "Day2"],
    "Return": [0.02, 0.01, 0.04, -0.01, 0.03, 0.02],
    "Volume": [200, 400, 890, 500, 350, 780]

})

df
```

```python
# 1) apply() — apply a function

df["Return_Pct"] = df["Return"].apply(lambda x: x * 100)
```

Quant tip: Prefer vectorized operations like df["Return"] * 100 over apply() when possible; they're usually faster and clearer.

```python
# 2) query() — filter rows cleanly

df.query("Return > 0.02")
```

```python
df.query("Return > 0.02 and Volume > 400")
```

```python
# Equivalent boolean indexing:

df[(df["Return"] > 0.02) & (df["Volume"] > 400)]
```

Use query() for readable filters; boolean indexing is more flexible for complex expressions.

```python
# 3) Sorting and index management

df.sort_values("Date")                 # Sort by date
df.sort_values("Return", ascending=False)  # Highest return first
df.set_index("Date")                   # Date becomes index
df.reset_index()                       # Index becomes a column
df.rename(columns={"Return": "Returns"})  # Rename a column
```
